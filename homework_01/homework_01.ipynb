## Задание 2: Подсчет выходных до отпуска

Пользователь вводит сколько дней осталось до ближайшего отпуска. Программа должна вывести количество выходных дней до отпуска, если учесть, что выходные это суббота и воскресенье, сегодня понедельник и праздники мы не учитываем.  

<br>

**Примеры:**
* `4  -> 0`
* `6  -> 1`
* `14 -> 4`

```python
count_days = int(input("Введите сколько дней осталось до отпуска"))
# наверно нужно посчитать кол-во полных недель (+2 за каждую неделю) и остаток (если == 6 то +1)

count_week = count_days//7
count_last_days = count_days%7

# суббота? тогда +1
is_need_add_plus_one = count_last_days == 6
addition = 0

if is_need_add_plus_one:
    addition = 1

result = count_week*2 + addition
print(result)
```

## Задание 3: Плитка шоколада

Пользователь вводит длину и ширину плитки шоколада, а также размер куска, который хочет отломить. Программа должна вычислить - можно ли совершить подобный разлом или нет, если учесть, что ломать можно только по прямой.

<br>

**Примеры:**
* `3, 4, 6  -> True`
* `5, 7, 8  -> False`
* `4, 5, 12 -> True`

```python
length = int(input("Введите длину плитки: "))
weight = int(input("Введите ширину плитки: "))
count_bite = int(input("Введите размер куска который отломать нужно: "))

# мозг сломан
# проверка что кусок не больше всей плитки

is_not_valid_size = count_bite > length * weight
if is_not_valid_size:
    print(False)  # кусок слишком большой
else:
    # проверяем разлом по прямой линии
    if count_bite % length == 0 and count_bite // length <= weight:
        print(True)
    elif count_bite % weight == 0 and count_bite // weight <= length:
        print(True)
    else:
        print(False)
```

## Задание 4: Римские числа

Пользователь вводит целое положительное число, программа должна вернуть строку в виде римского числа.

<br>

**Примеры:**
* `3   -> III`
* `15  -> XV`
* `234 -> CCXXXIV`

```python
# эээээээээээээээээээээээээ

# Римские числа:
# I=1, V=5, X=10, L=50, C=100, D=500, M=1000
# Повторяем символ до 3 раз: III=3, XX=20
# Если меньший слева от большего — вычитаем: IV=4, IX=9, XL=40, CM=900
# Иначе складываем слева направо: VIII=8, LX=60

number = int(input("Введите число (до 3999): "))
result = ''

### проверка на M - 1000
count_M = number // 1000
result += 'M' * count_M
remainder = number % 1000

### проверка на CM - 900
count_CM = remainder // 900
result += 'CM' * count_CM
remainder = remainder % 900

### проверка на D - 500
count_D = remainder // 500
result += 'D' * count_D
remainder = remainder % 500

### проверка на CD - 400
count_CD = remainder // 400
result += 'CD' * count_CD
remainder = remainder % 400

# C - 100
count_C = remainder // 100
result += 'C' * count_C
remainder = remainder % 100

# XC - 90
count_XC = remainder // 90
result += 'XC' * count_XC
remainder = remainder % 90

# L - 50
count_L = remainder // 50
result += 'L' * count_L
remainder = remainder % 50

# XL - 40
count_XL = remainder // 40
result += 'XL' * count_XL
remainder = remainder % 40

# X - 10
count_X = remainder // 10
result += 'X' * count_X
remainder = remainder % 10

# IX - 9
count_IX = remainder // 9
result += 'IX' * count_IX
remainder = remainder % 9

# V - 5
count_V = remainder // 5
result += 'V' * count_V
remainder = remainder % 5

# IV - 4
count_IV = remainder // 4
result += 'IV' * count_IV
remainder = remainder % 4

# I - 1
count_I = remainder
result += 'I' * count_I

print(result)
```

## Задание 5: Проверка на вещественное число

Пользователь вводит данные, проверить - являются ли они положительным вещественным числом. Не использовать встроенные функции для проверки, только методы данных и конструкцию IF. (Дополнительное задание, по желанию - проверка на отрицательные вещественные числа)

<br>

**Примеры:**
* `5.6  -> True`
* `.78  -> True`
* `.67. -> False`
* `5    -> True`

```python
######### yo soy francesco virgolini
### uiiiiiiiiiiiiiiiiii

input_data = input("Введите данные:")

dot_count = input_data.count('.')

if dot_count > 1:
    print(False)
else:
  # убираем точку
    temp = input_data.replace('.', '')

    # проверка на пустую строку
    if len(temp) == 0:
        print(False)
    # проверка на недопустимые символы
    elif temp.replace('0','').replace('1','').replace('2','').replace('3','')\
             .replace('4','').replace('5','').replace('6','').replace('7','')\
             .replace('8','').replace('9','') != '':
        print(False)
    else:
        print(True)

```
